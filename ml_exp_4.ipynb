```python
# Load dataset from local storage if 'df' is not defined
import pandas as pd
import io
from google.colab import files

if 'df' not in locals() or not isinstance(df, pd.DataFrame):
    print("Attempting to load 'titanic.csv'. If not found, you may be prompted to upload.")
    try:
        df = pd.read_csv("titanic.csv")
        print("Dataset loaded from 'titanic.csv' (existing file).")
    except FileNotFoundError:
        print("The file 'titanic.csv' was not found. Please upload the 'titanic.csv' file.")
        uploaded = files.upload() # This will prompt the user

        # Check for both titanic.csv and titanic (1).csv as a fallback
        if 'titanic.csv' in uploaded:
            df = pd.read_csv(io.BytesIO(uploaded['titanic.csv']))
            print("Dataset loaded as 'titanic.csv' from uploaded files.")
        elif 'titanic (1).csv' in uploaded:
            df = pd.read_csv(io.BytesIO(uploaded['titanic (1).csv']))
            print("Dataset loaded as 'titanic (1).csv' from uploaded files.")
        else:
            raise FileNotFoundError("No 'titanic.csv' or 'titanic (1).csv' was uploaded. Please ensure the correct file is uploaded.")
else:
    print("DataFrame 'df' already exists, skipping load.")

# Check for missing values
print("\nMissing values before preprocessing:\n", df.isnull().sum())
```

```python
# Fill missing values in 'Age' with the median age
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
## Fill missing 'Embarked' values with the most common value (mode)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
## Drop the 'Cabin' column because too many values are missing
df.drop(columns=['Cabin'], inplace=True)
```

```python
# Step 4: Check missing values after preprocessing
print("\nMissing values after preprocessing:\n", df.isnull().sum())
```

```python
print("\nDataset Information After Preprocessing:")
print(df.info())
```

```python
# Check for duplicates
print("Duplicate rows before removal:", df.duplicated().sum())
# Remove duplicate rows
df.drop_duplicates(inplace=True)
# Check for duplicates after removal
print("Duplicate rows after removal:", df.duplicated().sum())
```

```python
# Convert categorical column 'Sex' into numerical values
print("Unique values in 'Sex' before encoding:", df['Sex'].unique())
```

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
print("Unique values in 'Sex' after encoding:", df['Sex'].unique())
```

```python
scaler = MinMaxScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])
print("Age and Fare after scaling:\n", df[['Age', 'Fare']].head())
```
